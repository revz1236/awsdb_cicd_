```sql
MERGE INTO dev_ext.bronze.emp t
USING dev_ext.bronze.emp_updates s
ON t.emp_id = s.emp_id
WHEN MATCHED THEN
UPDATE SET 
  t.salary = s.salary,
  t.is_active = 'Y'
WHEN NOT MATCHED THEN
INSERT (
  emp_id,
  name,
  dept,
  salary,
  is_active
) 
VALUES (
  s.emp_id,
  s.name,
  s.dept,
  s.salary,
  'Y'
)
WHEN NOT MATCHED BY SOURCE THEN
UPDATE SET
t.is_active = 'N'
;
```
