```python
_dept = dbutils.widgets.get("dept")
```

```python
print(_dept)
```

```python
df = spark.read.csv("dbfs:/databricks-datasets/retail-org/company_employees/company_employees.csv", header=True)
```

```python
display(df)
```

```python
df_filtered = df.where(f"upper(department) = '{_dept.upper()}'")
display(df_filtered)
df_filtered.count()
```

```python
_count = df_filtered.count()
if _count > 0:
  df_filtered.write.mode("overwrite").saveAsTable(f"dev_ext.bronze.dept_{_dept}")
  print(f"Inserted {_count} records for {_dept}")
  display(spark.table(f"dev_ext.bronze.dept_{_dept}"))
else:
  print(f"No records found for {_dept}")
```

```python
dbutils.notebook.exit(_count)
```
